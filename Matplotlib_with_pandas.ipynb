# Doing pandas with matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Pokemon.csv")
df.info()
print(df.head(10).to_string(index=False))
type_count = (df["Type1"].value_counts())
print(type_count)
plt.barh(type_count.index,type_count.values,color="yellow",edgecolor="Black")
plt.title("Type Count",color="Green", size=15)
plt.xlabel("Count",color="Green", size=15)
plt.ylabel("PokeType",color="Green", size=15)
plt.tight_layout()
plt.show()
```
