```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# Let's just do students study hour vs their test score

```python
x = np.array([27,34,40,56,39,70,77,83,89,100])
y1 = np.array([0,1,2,3,3,4,5,6,7,8])
y2 = np.array([2,3,5,5,6,7,6,8,9,8])
my_fun = dict(color="Green",
            alpha= 0.7,
            s = 100,

            )
plt.title("Student Status",color="Red",size=20)
plt.xlabel("Marks",color="Red",size=10)
plt.ylabel("Study Time",color="Red",size=10)
plt.scatter(x,y1,**my_fun,marker="s",label="Class A")
plt.scatter(x,y2,**my_fun,label="Class B")
plt.legend()
plt.show()
```

# HISTOGRAM

```python
# Histogram is a visual representation of the distribution os qantitative data.
# They group together values into bins(intervals) and counts how many falls in each range
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.random.normal(loc=80,scale =10,size=100)
x = np.clip(x, 0,100)
plt.title("Grades",color="Red",size=20)
plt.xlabel("Marks",color="Red",size=10)
plt.ylabel("Frequency",color="Red",size=10)


plt.hist(x,color="Yellow",bins=15,
         edgecolor="Black")

plt.show()
```
