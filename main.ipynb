```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([2023,2024,2025,2026])
y = np.array([15,30,167,78])

plt.plot(x,y, marker="+",
        markersize = 10, # {ms } cna be used as {markerstyle}.
        markerfacecolor="cyan",  #{mfs} can be used as {markerfacecolor}.
        markeredgecolor="green", #{mec} can be used as {markeredgecolor}.
        linestyle="dotted", #{ls} can be used as {linestyle}.
        linewidth=5, # {lw} cna be used as {linewidth}
        color="pink")  #WE can use 'mfs' for 'markerfacecolour' and{ms} for {markersize}

# how to show our plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([2023,2024,2025,2026])
z = np.array([2027,2028,2029,2030])
y1 = np.array([15,30,167,78])
y2 = np.array([30,70,150,90])
y3 = np.array([60,90,120,90])

my_style = dict(marker="+",
        ms = 10, # {ms } cna be used as {markerstyle}.
        mfc="cyan",  #{mfs} can be used as {markerfacecolor}.
        mec="red", #{mec} can be used as {markeredgecolor}.
        ls="solid", #{ls} can be used as {linestyle}.
        lw=5, # {lw} cna be used as {linewidth}
        )  #WE can use 'mfs' for 'markerfacecolour' and{ms} for {markersize}

plt.title("Class Size", fontsize=20,
          family="Arial",
          fontweight="bold",
          color="red",
          )

plt.ylabel("Student count", fontsize=20,
          family="Arial",
          fontweight="bold",
          color="red",
          )

plt.xlabel("Year", fontsize=20,
          family="Arial",
          fontweight="bold",
          color="red",
          )


plt.plot(x,y1,color="yellow", **my_style )
plt.plot(x,y2,color="silver", **my_style )
plt.plot(x,y3,color="blue", **my_style )
plt.plot(z,y2,color="black", **my_style )
plt.plot(z,y1,color="green", **my_style )
plt.plot(z,y3,color="purple", **my_style )


#Parameters for ticks in x and y axis{change:colors, size etc}
plt.tick_params(axis="both",
                colors="red")

plt.xticks(x,z)

# how to show our plot
plt.show()
```

Grid lines

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array(["A", "B", "C", "D", "E"])
y = np.array([24, 45, 56, 67, 89])

plt.plot(
    x, y,
    marker="*",
    markersize=20,
    markerfacecolor="cyan",
    markeredgecolor="red",
    color="blue"
)

plt.grid(True)

plt.title("Test Result", fontsize=10, color="green", fontname="Arial")
plt.xlabel("Students", fontsize=10, color="green", fontname="Arial")
plt.ylabel("Marks", fontsize=10, color="green", fontname="Arial")

plt.tick_params(axis="both", colors="purple", labelsize=10)

plt.show()
```
