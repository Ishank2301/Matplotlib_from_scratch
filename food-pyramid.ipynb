# Here We have taken the example of food pyramid to plot a bar chart to better understand matplotlib


Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
Categories = np.array(["Grains","Fruits","Vegetables","protein","Dairy","Sweet"])
values = np.array([4,3,2,5,3,1])

plt.bar(Categories, values,color="Green",edgecolor="Black")
plt.title("Food Pyramid", color="Blue",size=10,font="Arabic")
plt.ylabel("Qantity", color="Blue",size=10,font="Arabic")
plt.xlabel("Food Type", color="Blue",size=10,font="Arabic")
plt.tick_params(axis="both", color="#ed4cda",labelsize=10,)

plt.show()
```

Import Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Implementing the bar Chart:



```python
categories = np.array(["Freshman","Sophomores","Juniours", "seniours"])
colorsss = np.array(["Green","Purple","Blue", "Red"])
values = np.array([400,300,200,500])

plt.pie( values,labels=Categories,
        autopct="%1.1f%%",
        colors=colorsss,
        explode=[0,0,0,0.2],
        shadow=True,
        startangle=180)
plt.title("Student Info",color="Green",size=10)

plt.show()
```
