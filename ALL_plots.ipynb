# Load the Csv File and perform(EDA)


```python
df = pd.read_csv("D:\AI_ML from scratch\Matplotlib_from_scratch\Students Social Media Addiction.csv")
df.info()
df.describe()
df.isnull().sum()
df.head()
print(df.iloc[0:10,0:9].to_string())
new_df = df["Country"].replace("India"=="Bharat")
```

# BARh graph:

```python
import matplotlib.pyplot as plt

usage = (
    df.groupby("Most_Used_Platform")["Avg_Daily_Usage_Hours"]
      .mean()
      .sort_values()
)
plt.figure(figsize=(8, 5))

plt.barh(
    usage.index,
    usage.values,
    color="red",
    edgecolor="yellow"
)

plt.title("Average Daily Usage by Social Media Platform", color="red", fontsize=15)
plt.xlabel("Time (Hours)", color="Black", fontsize=12)
plt.ylabel("App", color="Black", fontsize=12)

plt.show()



```

### Plot graph(Most used platform ):

```python
import matplotlib.pyplot as plt

usage = (
    df.groupby("Most_Used_Platform")["Avg_Daily_Usage_Hours"]
      .max()
      .sort_values().head(5)
)

usage1 = (
    df.groupby("Most_Used_Platform")["Addicted_Score"]
      .max()
      .sort_values().head(5)
)


plt.figure(figsize=(8, 5))

plt.plot(
    usage.index,
    usage.values,
    marker="*",
    color="red",
    markeredgecolor="yellow"
)

plt.plot(
    usage.index,
    usage.values,
    marker=".",
    markersize=20,
    lw=5,
    color="Green",
    markeredgecolor="Blue"
)


plt.plot(
    usage1.index,
    usage1.values,
    marker="*",
    markersize=20,
    color="red",
    lw=5,
    markeredgecolor="yellow"
)

plt.title("Social Media Platform Average Daily Usage", color="red", fontsize=15)
plt.xlabel("Apps", color="Black", fontsize=12)
plt.ylabel("Time (Hours)", color="Black", fontsize=12)

plt.show()
```

# Bar graph(count vs avg daily usage)


```python
#if i have to only access the top 10 country:
countries = (
    df.groupby("Country")["Avg_Daily_Usage_Hours"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)
plt.barh(countries.index,countries.values,color="Yellow",edgecolor="Black")
plt.title("country vs avg daily usage", color="Black",fontsize=15)
plt.xlabel("Time (Hours)", color="Black", fontsize=12)
plt.ylabel("countries", color="Black", fontsize=12)
plt.show()

```

# Piechart Top 10 country by daily usage

```python
countries = (
    df.groupby("Country")["Avg_Daily_Usage_Hours"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(6, 6))
plt.pie(
    countries.values,
    labels=countries.index,
    autopct="%1.1f%%",
    shadow=True,
    startangle=90,
    explode=[0,0,0.2,0,0,0,0,0,0,0]
)
plt.title("Top 10 Countries by Avg Daily Usage")
plt.show()
```
